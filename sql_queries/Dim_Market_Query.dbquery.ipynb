```sql
CREATE OR REPLACE TABLE gold.dim_market (
    market_key BIGINT GENERATED ALWAYS AS IDENTITY,
    market_symbol STRING,
    base_asset_key BIGINT,
    quote_currency STRING,
    exchange_name STRING,
    market_type STRING,
    is_active BOOLEAN
)
USING DELTA;


INSERT INTO gold.dim_market (
    market_symbol,
    base_asset_key,
    quote_currency,
    exchange_name,
    market_type,
    is_active
)
SELECT
    CONCAT(symbol, '-USD') AS market_symbol,
    asset_key AS base_asset_key,
    'USD' AS quote_currency,
    'Coinbase' AS exchange_name,
    'Spot' AS market_type,
    TRUE AS is_active
FROM gold.dim_asset;
```
