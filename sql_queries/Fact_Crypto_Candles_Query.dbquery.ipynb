```sql
CREATE OR REPLACE TABLE gold.fact_crypto_candles
USING DELTA
AS
SELECT
    s.candle_key,
    m.market_key,
    a.asset_key,
    d.date_key,
    e.exchange_key,
    s.product_id,
    s.base_asset_symbol,
    s.quote_asset_symbol,
    s.trade_timestamp,
    s.trade_date,
    s.open_price,
    s.high_price,
    s.low_price,
    s.close_price,
    s.volume,
    s.source_system,
    s.source_ingested_at,
    s.silver_processed_at
FROM silver.crypto_candles_clean s
LEFT JOIN gold.dim_market m
    ON s.product_id = m.market_symbol
    AND m.is_active = TRUE
LEFT JOIN gold.dim_asset a
    ON s.base_asset_symbol = a.symbol
    AND a.is_active = TRUE
LEFT JOIN gold.dim_date d
    ON s.trade_date = d.full_date
LEFT JOIN gold.dim_exchange e
    ON s.source_system = e.exchange_name
    AND e.is_active = TRUE;

SELECT COUNT(*) AS fact_row_count FROM gold.fact_crypto_candles;
```
